```python
timetagdf = pd.read_table(SampleName+".aligned.sorted.merged.tagged.time_tags.csv", sep = "\t")
human_df=pd.read_table(SampleName+".aligned.sorted.merged.tagged.HUMAN_ONLY.dge.txt.gz")
human_vector=human_df.iloc[:,1:].sum()
mouse_df=pd.read_table(SampleName+".aligned.sorted.merged.tagged.MOUSE_ONLY.dge.txt.gz")
mouse_vector=mouse_df.iloc[:,1:].sum()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the data
df = pd.read_csv('Species_Mixing_data.time_tags.csv')

# Define the three time tags and their labels
time_tags = ['SEQ10_GGTT', 'SEQ11_GGCC', 'SEQ12_GGGG']
labels = {
    'SEQ10_GGTT': 'Reservoir 1 \n50:50 Human(K562):Mouse(EL4)',
    'SEQ11_GGCC': 'Reservoir 2 \nHuman(K562)',
    'SEQ12_GGGG': 'Reservoir 3 \nMouse(EL4)'
}

# Create subdirectory for individual plots if it doesn't exist
os.makedirs('individual_time_tag_plots', exist_ok=True)

# First, create and display the combined plot
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, time_tag in enumerate(time_tags):
    ax = axes[i]
    
    # Filter data for this time tag
    df_tag = df[df['FINAL TIME TAG'] == time_tag]
    
    # Plot all points without categorization or colors
    ax.scatter(df_tag['HUMAN_COUNTS'], df_tag['MOUSE_COUNTS'], s=10, color='black')
    
    # Set limits and aspect
    max_count = max(df_tag['HUMAN_COUNTS'].max(), df_tag['MOUSE_COUNTS'].max()) if not df_tag.empty else 1000
    axis_limit = max_count * 1.1
    ax.set_xlim(0, axis_limit)
    ax.set_ylim(0, axis_limit)
    ax.set_aspect('equal')
    
    # Diagonal line
    ax.plot([0, axis_limit], [0, axis_limit], 'k--', alpha=0.3, linewidth=1)
    
    # Simple labels
    ax.set_xlabel('Human Counts')
    ax.set_ylabel('Mouse Counts')
    ax.set_title(f"Coinjected with Time-Tag {time_tag}")
    
    # Add smaller, more translucent text box higher in upper half, bold text
    ax.text(0.5, 0.75, labels[time_tag], transform=ax.transAxes, ha='center', va='center',
            fontsize=9, fontweight='bold', bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow', alpha=0.5, edgecolor='black'))

plt.tight_layout()

# Save combined plot as SVG and PNG in current directory
plt.savefig('combined_time_tags.svg', format='svg', bbox_inches='tight')
plt.savefig('combined_time_tags.png', format='png', bbox_inches='tight', dpi=300)

plt.show()

# Now, create and export individual plots as SVG
for time_tag in time_tags:
    # Create individual figure
    fig, ax = plt.subplots(1, 1, figsize=(5, 5))
    
    # Filter data for this time tag
    df_tag = df[df['FINAL TIME TAG'] == time_tag]
    
    # Plot all points without categorization or colors
    ax.scatter(df_tag['HUMAN_COUNTS'], df_tag['MOUSE_COUNTS'], s=10, color='black')
    
    # Set limits and aspect
    max_count = max(df_tag['HUMAN_COUNTS'].max(), df_tag['MOUSE_COUNTS'].max()) if not df_tag.empty else 1000
    axis_limit = max_count * 1.1
    ax.set_xlim(0, axis_limit)
    ax.set_ylim(0, axis_limit)
    ax.set_aspect('equal')
    
    # Diagonal line
    ax.plot([0, axis_limit], [0, axis_limit], 'k--', alpha=0.3, linewidth=1)
    
    # Simple labels
    ax.set_xlabel('Human Counts')
    ax.set_ylabel('Mouse Counts')
    ax.set_title(f"Coinjected with Time-Tag {time_tag}")
    
    # Add smaller, more translucent text box higher in upper half, bold text
    ax.text(0.5, 0.75, labels[time_tag], transform=ax.transAxes, ha='center', va='center',
            fontsize=9, fontweight='bold', bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow', alpha=0.5, edgecolor='black'))
    
    # Save as SVG in subdirectory
    filename = f'{time_tag}.svg'
    filepath = os.path.join('individual_time_tag_plots', filename)
    plt.savefig(filepath, format='svg', bbox_inches='tight')
    plt.close(fig)  # Close to avoid display
```
